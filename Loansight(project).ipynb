```python
from sklearn.preprocessing import LabelEncoder
```

```python
data['person_gender'] = a.fit_transform(data['person_gender'])
```

```python
a = LabelEncoder()
```

```python
data['loan_intent'] = a.fit_transform(data['loan_intent'])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
x = data.drop('loan_status',axis=1)
y = data['loan_status']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from collections import Counter
```

```python
Counter(y_train)
```

```python
from imblearn.over_sampling import SMOTE
```

```python
a = SMOTE()
```

```python
x_train_smt,y_train_smt = a.fit_resample(x_train,y_train)
```

```python
Counter(y_train_smt)
```

LOGISTIC REGRESSION

```python
model1 = LogisticRegression()
model1.fit(x_train_smt,y_train_smt)
```

```python
y_pred1 = model1.predict(x_test)
y_pred1
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
ac1 = accuracy_score(y_test,y_pred1)
ac1
```

DECISION TREE

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model2 = DecisionTreeClassifier(max_depth=5)
model2.fit(x_train_smt,y_train_smt)
```

```python
y_pred2 = model2.predict(x_test)
y_pred2
```

```python
ac2 = accuracy_score(y_test,y_pred2)
ac2
```

RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model3 = RandomForestClassifier(n_estimators=100)
model3.fit(x_train_smt,y_train_smt)
```

```python
y_pred3 = model3.predict(x_test)
y_pred3
```

```python
ac3 = accuracy_score(y_test,y_pred3)
ac3
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model4 = KNeighborsClassifier()
model4.fit(x_train_smt,y_train_smt)
```

```python
y_pred4 = model4.predict(x_test)
y_pred4
```

```python
ac4 = accuracy_score(y_test,y_pred4)
ac4
```
